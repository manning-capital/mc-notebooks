```python
import os

import pandas as pd
from dotenv import load_dotenv
from mc_postgres_db.models import (
    Asset,
)
from mc_postgres_db.operations import set_data
from sqlalchemy import create_engine, select

load_dotenv()

CMC_PRO_API_KEY = os.getenv("CMC_PRO_API_KEY")
POSTGRES_URL = os.getenv("POSTGRES_URL")

engine = create_engine(POSTGRES_URL)
```

```python
df = pd.read_sql(select(Asset), engine)
df["symbol"] = df["name"]
df["updated_at"] = pd.Timestamp.now()
df
```

```python
set_data(engine, Asset.__tablename__, df, operation_type="upsert")
```
