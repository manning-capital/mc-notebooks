```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sqrt_numerical(x: float | np.ndarray, max_iter: int = 20) -> float | np.ndarray:
    """
    Compute the square root of x.
    """

    def f(y: float | np.ndarray) -> float | np.ndarray:
        return y * y - x

    def df_dy(y: float | np.ndarray) -> float | np.ndarray:
        return 2 * y

    y = x / 2
    for _ in range(max_iter):
        y = y - f(y) / df_dy(y)

    return y


def sqrt_analytical(x: float) -> float:
    return x**0.5


x = np.linspace(0.001, 100, 1000)
y_numerical = sqrt_numerical(x)
y_analytical = sqrt_analytical(x)

plt.plot(x, y_analytical, label="Analytical")
plt.plot(x, y_numerical, label="Numerical")
plt.legend()
plt.show()
```
